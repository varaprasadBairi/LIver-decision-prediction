```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
cols = [
    "Age","Gender","Total_Bilirubin","Direct_Bilirubin",
    "Alkaline_Phosphotase","Alamine_Aminotransferase",
    "Aspartate_Aminotransferase","Total_Proteins",
    "Albumin","Albumin_Globulin_Ratio","Dataset"
]
```

```python
df = pd.read_csv("liver_patients.csv", header=None, names=cols)
```

```python
print("Shape of dataset:", df.shape)
print("\nColumns:")
print(df.columns)
```

```python

print(df.dtypes)
```

```python
df.head(5)
```

```python
print("\nTarget Class Count:")
print(df["Dataset"].value_counts())

plt.figure(figsize=(6,4))
df["Dataset"].value_counts().plot(kind="bar")
plt.title("Target Class Balance")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(df["Age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
corr = df.select_dtypes(include=np.number).corr()

plt.figure(figsize=(10,6))
plt.imshow(corr, interpolation='nearest')
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(df["Total_Bilirubin"].dropna())
plt.title("Boxplot - Total Bilirubin")
plt.ylabel("Values")
plt.show()
```

```python
df.hist(figsize=(14,10))
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(6,4))
df["Gender"].value_counts().plot(kind="bar")
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
for label in df["Dataset"].unique():
    subset = df[df["Dataset"] == label]
    plt.hist(subset["Age"], alpha=0.5, label=f"Class {label}")

plt.title("Age by Dataset Class")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```
